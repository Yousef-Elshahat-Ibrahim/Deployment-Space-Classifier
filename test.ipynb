```python
"""
Run this in the SAME Python environment you used to build/test the pipeline
(the one that has tensorflow, keras, alerce, sfdmap, etc. installed).

It pins exact versions for just the packages this project actually imports,
instead of dumping your whole environment with `pip freeze`.
"""

import importlib.metadata as md

# import name -> pip package name (they differ for a few of these)
PACKAGES = {
    "streamlit": "streamlit",
    "pandas": "pandas",
    "numpy": "numpy",
    "requests": "requests",
    "tensorflow": "tensorflow",
    "keras": "keras",
    "sklearn": "scikit-learn",
    "sfdmap": "sfdmap",
    "astropy": "astropy",
    "dust_extinction": "dust_extinction",
    "alerce": "alerce",
}

lines = []
missing = []

for import_name, pip_name in PACKAGES.items():
    try:
        version = md.version(pip_name)
        lines.append(f"{pip_name}=={version}")
    except md.PackageNotFoundError:
        missing.append(pip_name)

with open("requirements.txt", "w") as f:
    f.write("\n".join(lines) + "\n")

print("Wrote requirements.txt:")
print("\n".join(lines))

if missing:
    print("\nWARNING: these packages were not found in this environment "
          "(not installed, or run this script from the wrong venv):")
    for m in missing:
        print(f"  - {m}")
```
